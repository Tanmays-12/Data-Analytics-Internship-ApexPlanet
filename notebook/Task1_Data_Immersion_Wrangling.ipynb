```python
import pandas as pd
```

# Step 1: Data Access & Familiarization

### Purpose

In this step, the dataset is loaded into a Pandas DataFrame to understand its structure before performing any analysis or cleaning.

The following information will be examined:

- Dataset preview
- Number of rows and columns
- Column names
- Data types
- Dataset summary

```python
df = pd.read_excel("../data/ApexPlanet_DataAnalytics_Dataset.xlsx")
```

## Dataset Exploration

After loading the dataset, an initial exploration is performed to understand its structure and contents. This includes previewing the data, checking the dataset dimensions, examining column names, and reviewing data types along with non-null values.

```python
# Preview the dataset
display(df.head())

# Dataset dimensions
print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nDataset Information:")
df.info()
```

# Step 2: Data Quality Assessment

### Objective

The dataset is assessed to identify potential data quality issues before cleaning.

The following checks are performed:

- Missing Values
- Duplicate Records
- Inconsistent Formatting
- Outlier Detection

```python
# Missing Values
print("Missing Values:")
display(df.isnull().sum().to_frame(name="Missing Values"))

# Duplicate Records
print("\nDuplicate Records:")
print(df.duplicated().sum())

# Inconsistent Formatting
print("\nUnique Values")

print("\nGender")
display(df["Gender"].unique())

print("\nCity")
display(df["City"].unique())

print("\nProduct")
display(df["Product"].value_counts())

print("\nCategory")
display(df["Category"].value_counts())

# Summary Statistics
print("\nSummary Statistics")
display(df.describe())
```

# Step 3: Data Cleaning & Transformation

### Objective

The identified data quality issues are resolved by handling missing values, removing duplicate records, and standardizing the date format.

```python
# Handle Missing Values
df["Age"] = df["Age"].fillna(df["Age"].median())
df["City"] = df["City"].fillna(df["City"].mode()[0])

print("Missing Values After Cleaning")
display(df.isnull().sum().to_frame(name="Missing Values"))

# Remove Duplicates
df.drop_duplicates(inplace=True)

print("\nDuplicate Records Remaining")
print(df.duplicated().sum())

# Standardize Date
df["Order_Date"] = pd.to_datetime(df["Order_Date"])

print("\nOrder_Date Data Type")
display(df["Order_Date"].dtype)
```

# Step 4: Feature Engineering

### Objective

A new feature named **High_Value_Order** is created to classify transactions based on Total Sales.

**Business Rule**

- Total Sales ≥ ₹100000 → Yes
- Otherwise → No

```python
df["High_Value_Order"] = df["Total_Sales"].apply(
    lambda x: "Yes" if x >= 100000 else "No"
)

display(df[["Total_Sales", "High_Value_Order"]].head())
```

# Step 5: Export Cleaned Dataset

### Objective

The cleaned dataset is exported as an Excel file for further analysis and future business reporting.

```python
df.to_excel("../data/cleaned_dataset.xlsx", index=False)

print(" Cleaned dataset exported successfully.")
print("Location: ../data/cleaned_dataset.xlsx")
```
